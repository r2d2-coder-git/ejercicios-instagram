# 🐼 Pandas: Junior vs Senior
Mismo resultado, código muy distinto.

### 💼 SUMA LAS VENTAS
### DE LA CATEGORÍA 'TECH'

```python
import pandas as pd

df = pd.DataFrame({
  'producto': ['A','B','C','D'],
  'categoria': ['tech','hogar',
                'tech','hogar'],
  'ventas': [100, 50, 200, 75],
})
```

```python
# JUNIOR ❌ recorre fila a fila
total = 0
for i, fila in df.iterrows():
    if fila['categoria'] == 'tech':
        total += fila['ventas']

print(total)
```

```python
# SENIOR ✅ máscara vectorizada
total = df.loc[
    df['categoria'] == 'tech',
    'ventas'
].sum()

print(total)
```

### 💼 CREA UNA COLUMNA:
### 'top' SI VENTAS >= 100,
### SI NO 'normal'

```python
# JUNIOR ❌ apply fila a fila
def etiqueta(fila):
    if fila['ventas'] >= 100:
        return 'top'
    return 'normal'

df['tipo'] = df.apply(
    etiqueta, axis=1)

print(df['tipo'].tolist())
```

```python
# SENIOR ✅ np.where vectorizado
import numpy as np

df['tipo'] = np.where(
    df['ventas'] >= 100,
    'top', 'normal')

print(df['tipo'].tolist())
```

### 💼 CUENTA CUÁNTOS
### PRODUCTOS HAY POR
### CATEGORÍA

```python
# JUNIOR ❌ diccionario a mano
conteo = {}
for cat in df['categoria']:
    conteo[cat] = \
        conteo.get(cat, 0) + 1

print(conteo)
```

```python
# SENIOR ✅ value_counts
conteo = df['categoria'].value_counts()

print(conteo)
```

### 💼 CRUZA VENTAS CON
### PRECIOS Y CALCULA
### EL INGRESO

```python
ventas = pd.DataFrame({
  'producto': ['A','B','C'],
  'unidades': [10, 5, 8]})

precios = pd.DataFrame({
  'producto': ['A','B','C'],
  'precio': [2, 3, 4]})
```

```python
# JUNIOR ❌ doble bucle anidado
ingresos = []
for i, v in ventas.iterrows():
    for j, p in precios.iterrows():
        if v['producto'] == p['producto']:
            ingresos.append(
                v['unidades'] * p['precio'])

print(ingresos)
```

```python
# SENIOR ✅ merge
df2 = ventas.merge(
    precios, on='producto')
df2['ingreso'] = (
    df2['unidades'] * df2['precio'])

print(df2['ingreso'].tolist())
```

### 💼 SACA LOS 2 MÁXIMOS
### GOLEADORES DE CADA
### EQUIPO

```python
goles = pd.DataFrame({
  'equipo': ['A','A','A',
             'B','B','B'],
  'jugador': ['p1','p2','p3',
              'p4','p5','p6'],
  'goles': [3, 1, 5, 2, 8, 4]})
```

```python
# JUNIOR ❌ ordena y recorre
top = []
for equipo in goles['equipo'].unique():
    sub = goles[
        goles['equipo'] == equipo]
    sub = sub.sort_values(
        'goles', ascending=False)
    top.append(sub.head(2))
top = pd.concat(top)

print(top['jugador'].tolist())
```

```python
# SENIOR ✅ sort + groupby.head
top = (
    goles
    .sort_values('goles',
        ascending=False)
    .groupby('equipo')
    .head(2)
)

print(top['jugador'].tolist())
```

### 💼 MEDIA, MÁXIMO Y SUMA
### DE GOLES POR EQUIPO

```python
# JUNIOR ❌ tres groupby sueltos
media = goles.groupby(
    'equipo')['goles'].mean()
maximo = goles.groupby(
    'equipo')['goles'].max()
suma = goles.groupby(
    'equipo')['goles'].sum()

print(media.tolist())
print(maximo.tolist())
print(suma.tolist())
```

```python
# SENIOR ✅ un solo agg
resumen = (
    goles
    .groupby('equipo')['goles']
    .agg(['mean', 'max', 'sum'])
)

print(resumen)
```

### 💼 LIMPIA LA COLUMNA
### PRECIO ('1.200€') Y
### SUMA EL TOTAL

```python
dd = pd.DataFrame({
  'precio': ['1.200€', '950€',
             '2.500€']})
```

```python
# JUNIOR ❌ limpia fila a fila
total = 0
for i, f in dd.iterrows():
    limpio = (f['precio']
        .replace('€', '')
        .replace('.', ''))
    total += int(limpio)

print(total)
```

```python
# SENIOR ✅ .str vectorizado
total = (
    dd['precio']
    .str.replace(r'[€.]', '',
        regex=True)
    .astype(int)
    .sum()
)

print(total)
```
