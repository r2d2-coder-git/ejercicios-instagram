### **SEGURA? MÍNIMO 8, UNA**

### **MAYÚSCULA Y UN NÚMERO 🔒**

```python
contrasenas = ["hola123", "Python2024!", 
               "abc", "Segura99"]

def es_segura(clave: str) -> bool:
    return (len(clave) >= 8
            and any(c.isupper() for c in clave)
            and any(c.isdigit() for c in clave))

print([es_segura(c) for c in contrasenas])
```

**DETECTA LAS SUSCRIPCIONES**

**FANTASMA Y CUÁNTO TE**

**CUESTAN AL AÑO 👻💸**

```python
from collections import Counter

gastos = [
  ("Netflix", 12.99), ("Mercadona", 45.20),
  ("Netflix", 12.99), ("Spotify", 9.99),
  ("Gasolina", 60.0), ("Spotify", 9.99),
  ("Netflix", 12.99), ("Spotify", 9.99)]

def suscripciones(gastos: list) -> dict:
    conteo = Counter(gastos)
    return {
        concepto: round(importe * 12, 2)
        for (concepto, importe), veces
        in conteo.items()
        if veces >= 2}

print(suscripciones(gastos))
```

### **CALCULA TU RACHA MÁS**

### **LARGA DE DÍAS SEGUIDOS**

### **(ESTILO DUOLINGO) 🔥**

```python
from datetime import date

dias = ["2025-01-01", "2025-01-02",
        "2025-01-03", "2025-01-05",
        "2025-01-06"]

def racha_maxima(dias: list) -> int:
    fechas = sorted(date.fromisoformat(d)
                    for d in dias)
    mejor = actual = 1
    for i in range(1, len(fechas)):
        if (fechas[i]- fechas[i-1]).days == 1:
            actual += 1
            mejor = max(mejor, actual)
        else:
            actual = 1
    return mejor

print(f"Racha: {racha_maxima(dias)} días")
```

**AGRUPA LOS CLICS DE UN**

**USUARIO EN SESIONES**

**(HUECO > 30 MIN) 🖱️**

```python
from datetime import datetime

clics = ["10:00", "10:05", "10:20",
         "11:15", "11:20", "14:00"]

def contar_sesiones(horas: list,
                    hueco_max: int = 30) -> int:
    marcas = [datetime.strptime(h, "%H:%M")
              for h in horas]
    sesiones = 1
    for i in range(1, len(marcas)):
        hueco = (marcas[i] - marcas[i-1]
                 ).total_seconds() / 60
        if hueco > hueco_max:
            sesiones += 1
    return sesiones

print(f"{contar_sesiones(clics)} sesiones")
```

**DETECTA FRAUDE: 3+**

**COMPRAS EN MENOS DE**

**2 MINUTOS 🚨💳**

```python
from datetime import datetime

compras = ["12:00:01", "12:00:30",
           "12:00:50", "15:00:00"]

def hay_fraude(horas: list, n: int = 3,
               ventana: int = 120) -> bool:
    marcas = sorted(
        datetime.strptime(h, "%H:%M:%S")
        for h in horas)
    return any(
        (marcas[i+n-1] - marcas[i]
         ).total_seconds() <= ventana
        for i in range(len(marcas) - n + 1))

print(hay_fraude(compras))
```

**"QUIEN COMPRÓ ESTO**

**TAMBIÉN COMPRÓ..."**

**UN MINI RECOMENDADOR 🛒**

```python
from collections import Counter

cestas = [
  ["pan", "leche", "huevos"],
  ["pan", "leche"],
  ["pan", "cerveza"],
  ["leche", "huevos"]]

def tambien_compraron(cestas: list,
                      producto: str) -> tuple:
    c = Counter()
    for cesta in cestas:
        if producto in cesta:
            c.update(p for p in cesta
                     if p != producto)
    return c.most_common(1)[0]

print(tambien_compraron(cestas, "pan"))
```

**¿QUÉ HASHTAG ESTÁ**

**PETANDO? EL DE MAYOR**

**CRECIMIENTO % 📈**

```python
pasada = {"#python": 100, "#ia": 50,
          "#data": 200}
actual = {"#python": 150, "#ia": 300,
          "#data": 210}

def mas_viral(pasada: dict,
              actual: dict) -> str:
    crecimiento = {
        h: (actual[h] - pasada[h])
        / pasada[h] * 100
        for h in actual if h in pasada}
    return max(crecimiento,
               key=crecimiento.get)

print(mas_viral(pasada, actual))
```

**¿ERES UN BOT? DETECTA**

**SI TUS CLICS TIENEN**

**INTERVALOS PERFECTOS 🤖**

```python
clics = [0, 5, 10, 15, 20]  # segundos

def parece_bot(tiempos: list) -> bool:
    intervalos = {
        tiempos[i+1] - tiempos[i]
        for i in range(len(tiempos) - 1)}
    return len(intervalos) == 1

print(parece_bot(clics))
```
