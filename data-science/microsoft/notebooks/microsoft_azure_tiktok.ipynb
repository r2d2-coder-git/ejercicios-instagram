### 💼 PREGUNTA ENTREVISTA:
### ¿Qué servicio Azure
### factura más y con
### qué ticket medio?

El mix de servicios define el negocio.
Volumen alto no siempre = ticket alto.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/azure_cloud.csv'
)

r = (
  df
  .groupby('servicio')
  .agg(
    coste=('coste_usd', 'sum'),
    ticket=('coste_usd', 'mean'),
    n=('cliente_id', 'count'),
  )
  .round(2)
  .sort_values('coste',
    ascending=False)
)

print('Facturación por servicio:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué regiones cobran
### más caro por unidad
### de uso?

Coste unitario = coste / uso. Revela
el sobrecoste geográfico real.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/azure_cloud.csv'
)

g = (
  df
  .groupby('region')
  .agg(
    coste=('coste_usd', 'sum'),
    uso=('uso_unidades', 'sum'),
  )
)

g['coste_x_unidad'] = (
  g['coste'] / g['uso']
).round(4)

print('Coste unitario por región:')
print(
  g.sort_values('coste_x_unidad',
    ascending=False)
  .round(2)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuánto más gasta un
### cliente Enterprise que
### uno Pay-As-You-Go?

Comparamos gasto medio por nivel
para ver el valor del upsell.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/azure_cloud.csv'
)

g = (
  df
  .groupby(['nivel', 'cliente_id'])
  ['coste_usd']
  .sum()
  .groupby('nivel')
  .mean()
  .round(0)
)

base = g.get('Pay-As-You-Go', 1)
lift = (g / base).round(2)

print('Gasto medio por cliente:')
print(g)
print('\nMultiplicador vs PAYG:')
print(lift)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué sectores
### concentran el gasto?
### (Pareto 80/20)

¿Pocos sectores aportan la mayoría
del ingreso? Clave para priorizar.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/azure_cloud.csv'
)

g = (
  df
  .groupby('sector')
  ['coste_usd']
  .sum()
  .sort_values(ascending=False)
)

acum = (
  g.cumsum() / g.sum() * 100
).round(1)

tabla = pd.DataFrame({
  'coste': g.round(0),
  'acum_%': acum,
})

print('Pareto de gasto por sector:')
print(tabla)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿El gasto cloud crece
### mes a mes? (tendencia
### y % de variación)

Miramos el coste mensual total y
su crecimiento intermensual.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/azure_cloud.csv'
)

s = (
  df
  .groupby('mes')['coste_usd']
  .sum()
  .sort_index()
)

tabla = pd.DataFrame({
  'coste': s.round(0),
  'var_%': (
    s.pct_change() * 100
  ).round(1),
})

print('Tendencia mensual:')
print(tabla)
```
