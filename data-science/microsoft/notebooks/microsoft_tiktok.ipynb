### 💼 PREGUNTA ENTREVISTA:
### ¿Qué 5 juegos nos dan
### más dinero al año
### y qué % del total?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/xbox_game_pass.csv'
)

total = df['ingresos_usd'].sum()

top = (
  df
  .groupby('juego')['ingresos_usd']
  .sum()
  .sort_values(ascending=False)
  .head(5)
)

tabla = pd.DataFrame({
  'ingresos_M': (top / 1e6).round(1),
  '%_total': (top / total * 100).round(1),
})

print('Top 5 juegos por ingresos:')
print(tabla)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué género engancha
### más horas y cuál
### monetiza mejor?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/xbox_game_pass.csv'
)

r = (
  df
  .groupby('genero')
  .agg(
    horas=('horas_medias_jugador',
           'mean'),
    gasto=('gasto_medio_usd',
           'mean'),
    jugadores=('jugadores_activos',
               'sum'),
  )
  .round(2)
  .sort_values('gasto',
    ascending=False)
)

print('Por género:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿La campaña de Navidad
### mueve de verdad
### a los jugadores?

Comparamos el mes pico de dic
con la media del resto del año.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/xbox_game_pass.csv'
)

df['m'] = (
  df['mes'].str[-2:].astype(int)
)

por_mes = (
  df
  .groupby('m')['jugadores_activos']
  .sum()
)

dic = por_mes[12]
resto = por_mes.drop(12).mean()
lift = (dic / resto - 1) * 100

print(f'Diciembre: {dic/1e6:.1f}M')
print(f'Media resto: {resto/1e6:.1f}M')
print(f'Subida Navidad: +{lift:.1f}%')
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué estudio recién
### comprado (Activision,
### Blizzard, King) aporta más?

Microsoft compró Activision Blizzard
King. ¿Cuánto pesa en el catálogo?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/xbox_game_pass.csv'
)

abk = ['Activision', 'Blizzard',
       'King']

df['grupo'] = (
  df['estudio']
  .apply(lambda e:
    'Activision-Blizzard-King'
    if e in abk else 'Resto Xbox')
)

r = (
  df
  .groupby('grupo')['ingresos_usd']
  .sum()
)

r = (r / 1e6).round(1)
pct = (r / r.sum() * 100).round(1)

print('Ingresos (M USD):')
print(r)
print('\n% del total:')
print(pct)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué juegos retienen
### bien pero monetizan
### mal? (oportunidad)

Muchas horas + poco gasto = gente
enganchada a la que no sacamos \$.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/xbox_game_pass.csv'
)

g = (
  df
  .groupby('juego')
  .agg(
    horas=('horas_medias_jugador',
           'mean'),
    gasto=('gasto_medio_usd',
           'mean'),
  )
)

# Enganchan mucho (horas > mediana)
# pero gastan poco (gasto < mediana)
hm = g['horas'].median()
gm = g['gasto'].median()

oport = (
  g[(g['horas'] > hm)
    & (g['gasto'] < gm)]
  .sort_values('horas',
    ascending=False)
  .round(2)
)

print('Enganchan pero no pagan:')
print(oport)
```
