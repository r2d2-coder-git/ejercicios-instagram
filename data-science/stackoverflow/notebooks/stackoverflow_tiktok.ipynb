### 💼 PREGUNTA ENTREVISTA:
### ¿Python paga más
### que JavaScript?  (DONE)

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/survey_results_public.csv'
)

langs = ['Python', 'JavaScript',
         'Java', 'C#', 'TypeScript',
         'Go', 'Rust', 'SQL']

# Filtrar con salario válido
df = df[df['ConvertedCompYearly'].notna()]

salarios = {}
for lang in langs:
  mask = df[
    'LanguageHaveWorkedWith'
  ].str.contains(lang, na=False)
  salarios[lang] = (
    df[mask]
    ['ConvertedCompYearly']
    .median()
  )

result = pd.Series(salarios)
result = result.sort_values(
  ascending=False
)

print("Salario mediano por lenguaje:")
print(result.round(0))
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuánto cobra un dev
### remoto vs presencial? (TODO)

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/survey_results_public.csv'
)

df = df[
  df['ConvertedCompYearly'].notna()
  & df['RemoteWork'].notna()
]

result = (
  df
  .groupby('RemoteWork')
  ['ConvertedCompYearly']
  .agg(['median', 'count'])
  .sort_values('median',
    ascending=False)
)

print("Salario por modalidad:")
print(result.round(0))
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuántos años tardas
### en duplicar tu salario?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/survey_results_public.csv'
)

df = df[
  df['ConvertedCompYearly'].notna()
  & df['WorkExp'].notna()
]

bins = [0, 2, 5, 10, 15, 20, 50]
labels = [
  '0-2', '2-5', '5-10',
  '10-15', '15-20', '20+'
]

df['exp_range'] = pd.cut(
  df['WorkExp'],
  bins=bins, labels=labels
)

result = (
  df
  .groupby('exp_range')
  ['ConvertedCompYearly']
  .median()
)

print("Salario por experiencia:")
print(result.round(0))

ratio = (
  result['10-15'] / result['0-2']
)
print(f"\nx{ratio:.1f} en 10-15 años")
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué % de devs usa IA
### y cuánto cobran vs
### los que no la usan? (DONE)

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/survey_results_public.csv'
)

df = df[
  df['AISelect'].notna()
  & df['ConvertedCompYearly'].notna()
]

# Simplificar: usa IA vs no
df['usa_ia'] = df['AISelect'].apply(
  lambda x: 'No usa IA'
  if 'No' in str(x)
  else 'Usa IA'
)

result = (
  df
  .groupby('usa_ia')
  ['ConvertedCompYearly']
  .agg(['median', 'count'])
)

total = result['count'].sum()
result['%'] = (
  result['count'] / total * 100
)

print(result.round(0))
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué nivel de estudios
### tienen los devs mejor
### pagados?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/survey_results_public.csv'
)

df = df[
  df['ConvertedCompYearly'].notna()
  & df['EdLevel'].notna()
]

result = (
  df
  .groupby('EdLevel')
  ['ConvertedCompYearly']
  .agg(['median', 'count'])
  .sort_values('median',
    ascending=False)
)

# Solo los que tienen +100 resp
result = result[
  result['count'] > 100
]

print("Salario por estudios:")
print(result.round(0))
```
