### 💼 PREGUNTA ENTREVISTA:
### ¿Python paga más
### que JavaScript?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/survey_results_public.csv'
)

langs = ['Python', 'JavaScript',
         'Java', 'C#', 'TypeScript',
         'Go', 'Rust', 'SQL']

#Filtrar con salario válido
df = df[df['ConvertedCompYearly'].notna()]

#Calcular media de salario por lenguaje
salarios = {}

for lang in langs:
    mask = df[
        'LanguageHaveWorkedWith'
    ].str.contains(lang, na=False)

    salarios[lang] = (
        df[mask]['ConvertedCompYearly']
        .median()
    )

#Construir dataframe y ordenar
result = pd.Series(salarios)
result = result.sort_values(
    ascending=False
)

print("Salario mediano por lenguaje:")
print(result.round(0))
```

### ¿Qué % de devs usa IA
### y cuánto cobran vs
### los que no la usan?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/survey_results_public.csv'
)

# Limpiar nulos
df = df[
    df['AISelect'].notna()
    & df['ConvertedCompYearly'].notna()
]

# Columna calculada
df['usa_ia'] = df['AISelect'].apply(
    lambda x: 'No usa IA'
    if 'No' in str(x)
    else 'Usa IA'
)

# Calculo de salario medio
result = (
    df.groupby('usa_ia')
    ['ConvertedCompYearly']
    .agg(['mean', 'count'])
)

total = result['count'].sum()
result['%'] = (
    result['count'] / total * 100
)

print(result.round(0))
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuánto cobra un dev
### remoto vs presencial?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/survey_results_public.csv'
)

df = df[
  df['ConvertedCompYearly'].notna()
  & df['RemoteWork'].notna()
]

result = (
  df
  .groupby('RemoteWork')
  ['ConvertedCompYearly']
  .agg(['median', 'count'])
  .sort_values('median',
    ascending=False)
)

print("Salario por modalidad:")
print(result.round(0))
```
