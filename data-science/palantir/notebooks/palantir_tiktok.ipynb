### 💼 PREGUNTA ENTREVISTA:
### ¿Pesa más el negocio
### de Gobierno o el
### Comercial? (valor total)

Palantir nació del gobierno (Gotham)
y empuja ahora lo comercial (Foundry).

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/palantir_contratos.csv'
)

r = (
  df
  .groupby('tipo_cliente')
  .agg(
    valor=('valor_m_usd', 'sum'),
    n=('contrato_id', 'count'),
    ticket=('valor_m_usd', 'mean'),
  )
  .round(1)
)

r['%'] = (
  r['valor'] / r['valor'].sum() * 100
).round(1)

print('Gobierno vs Comercial:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué producto genera
### más valor y cuál
### tiene mejor margen?

Gotham, Foundry, Apollo y AIP (IA).
¿Dónde está el dinero de verdad?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/palantir_contratos.csv'
)

r = (
  df
  .groupby('producto')
  .agg(
    valor=('valor_m_usd', 'sum'),
    margen=('margen_bruto', 'mean'),
  )
  .round(2)
  .sort_values('valor',
    ascending=False)
)

print('Valor y margen por producto:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué tipo de cliente
### renueva más? (clave
### del modelo SaaS)

La renovación es la vida de Palantir:
contrato que no renueva es ingreso que cae.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/palantir_contratos.csv'
)

r = (
  df
  .groupby('tipo_cliente')
  ['renovado']
  .mean()
  .mul(100)
  .round(1)
  .sort_values(ascending=False)
)

print('Tasa de renovación (%):')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Contratos largos
### valen más por mes?
### (valor normalizado)

Valor / duración = cuánto factura
cada contrato al mes de verdad.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/palantir_contratos.csv'
)

df['valor_mes'] = (
  df['valor_m_usd']
  / df['duracion_meses']
)

r = (
  df
  .groupby('duracion_meses')
  ['valor_mes']
  .mean()
  .round(3)
)

print('Valor mensual por duración:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué sectores crecen
### en valor año
### tras año?

Pivotamos sector x año para ver la
tendencia de la cartera.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/palantir_contratos.csv'
)

r = pd.pivot_table(
  df,
  index='sector',
  columns='anio',
  values='valor_m_usd',
  aggfunc='sum',
  fill_value=0,
).round(0)

print('Valor por sector y año:')
print(r)
```
