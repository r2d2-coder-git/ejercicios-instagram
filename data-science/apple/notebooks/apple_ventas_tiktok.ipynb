### 💼 PREGUNTA ENTREVISTA:
### ¿Qué categoría genera
### más ingresos y qué
### precio medio tiene?

iPhone, iPad, Mac, Wearables...
¿Quién sostiene de verdad la caja?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/apple_ventas.csv'
)

r = (
  df
  .groupby('categoria')
  .agg(
    ingresos=('ingresos_usd', 'sum'),
    unidades=('unidades', 'sum'),
    precio=('precio_usd', 'mean'),
  )
  .round(0)
  .sort_values('ingresos',
    ascending=False)
)

print('Ingresos por categoría:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué modelos aportan
### el 80% de los
### ingresos? (Pareto)

Pocos modelos suelen cargar con casi
todo. Buscamos ese núcleo.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/apple_ventas.csv'
)

g = (
  df
  .groupby('modelo')
  ['ingresos_usd']
  .sum()
  .sort_values(ascending=False)
)

acum = (
  g.cumsum() / g.sum() * 100
).round(1)

tabla = pd.DataFrame({
  'ingresos_M': (g / 1e6).round(2),
  'acum_%': acum,
})

print('Pareto de modelos:')
print(tabla.head(8))
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué peso tiene cada
### región en las ventas
### de iPhone?

El reparto geográfico decide dónde
invertir en marketing y stock.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/apple_ventas.csv'
)

ip = df[df['categoria'] == 'iPhone']

g = (
  ip
  .groupby('region')
  ['ingresos_usd']
  .sum()
  .sort_values(ascending=False)
)

pct = (g / g.sum() * 100).round(1)

tabla = pd.DataFrame({
  'ingresos_M': (g / 1e6).round(2),
  '%': pct,
})

print('iPhone por región:')
print(tabla)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuánto caen las
### ventas de un modelo
### según envejece?

Antigüedad = año - año_lanzamiento.
Medimos el decay de unidades.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/apple_ventas.csv'
)

df['antig'] = (
  df['anio']
  - df['anio_lanzamiento']
)

r = (
  df
  .groupby('antig')
  ['unidades']
  .mean()
  .round(0)
)

r = r.to_frame('uds_medias')
r['vs_nuevo_%'] = (
  r['uds_medias']
  / r['uds_medias'].iloc[0] * 100
).round(1)

print('Decay por antigüedad:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿El Q4 (navidad)
### dispara las ventas
### frente al resto?

Comparamos ingresos medios por
trimestre del año (estacionalidad).

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/apple_ventas.csv'
)

df['q'] = (
  df['trimestre'].str[-2:]
)

g = (
  df
  .groupby('q')['ingresos_usd']
  .sum()
)

idx = (g / g.mean() * 100).round(1)

tabla = pd.DataFrame({
  'ingresos_M': (g / 1e6).round(2),
  'indice': idx,
})

print('Estacionalidad por trimestre:')
print(tabla)
```
