### 💼 PREGUNTA ENTREVISTA:
### ¿Subir de nivel mejora
### de verdad el rating
### del vendedor?

Rating medio por nivel. ¿Los Top
Rated lo son por algo?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/'
  'ebay_vendedores.csv'
)

orden = ['Nuevo', 'Estandar',
  'Destacado', 'PowerSeller',
  'Top Rated']

r = (
  df
  .groupby('nivel')
  .agg(
    rating=('rating', 'mean'),
    ventas=('ventas', 'mean'),
    n=('rating', 'size'),
  )
  .round(2)
  .reindex(orden)
)

print('Rating y ventas por nivel:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué país concentra a
### los mejores vendedores
### por volumen?

Ventas totales por país y su peso
sobre el total del marketplace.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/'
  'ebay_vendedores.csv'
)

r = (
  df
  .groupby('pais')['ventas']
  .sum()
  .sort_values(ascending=False)
)

pct = (r / r.sum() * 100).round(1)

tabla = pd.DataFrame({
  'ventas': r,
  '%': pct,
})

print('Ventas por país:')
print(tabla)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué nivel devuelve
### más pedidos en
### proporción?

Tasa devolución = devoluciones /
ventas. Devolver mata el margen.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/'
  'ebay_vendedores.csv'
)

g = (
  df
  .groupby('nivel')
  [['devoluciones', 'ventas']]
  .sum()
)

g['tasa_dev_%'] = (
  g['devoluciones']
  / g['ventas'] * 100
).round(2)

print('Tasa devolución por nivel:')
print(
  g['tasa_dev_%']
  .sort_values(ascending=False)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Más ventas significa
### peor rating? ¿Hay
### correlación?

Correlación ventas vs rating. El
mito de 'grande = peor trato'.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/'
  'ebay_vendedores.csv'
)

corr = (
  df['ventas']
  .corr(df['rating'])
)

df['tramo'] = pd.qcut(
  df['ventas'], 4,
  labels=['Q1', 'Q2',
          'Q3', 'Q4'],
)

r = (
  df
  .groupby('tramo',
    observed=True)['rating']
  .mean()
  .round(2)
)

print(f'Correlación: {corr:.3f}')
print('\nRating por cuartil ventas:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué vendedores son
### 'estrella': muchas
### ventas y top rating?

Filtro combinado: alto volumen +
rating >= 99. Los que escalar.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/'
  'ebay_vendedores.csv'
)

umbral = df['ventas'].quantile(.75)

estrella = df[
  (df['ventas'] >= umbral)
  & (df['rating'] >= 99)
]

pct = (
  len(estrella)
  / len(df) * 100
)

print(f'Umbral ventas: {umbral:.0f}')
print(f'Estrella: {len(estrella)}')
print(f'% del total: {pct:.1f}%')
print('\nPor país:')
print(
  estrella['pais']
  .value_counts()
)
```
