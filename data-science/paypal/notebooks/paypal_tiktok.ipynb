### 💼 PREGUNTA ENTREVISTA:
### ¿Qué tipo de pago
### mueve más volumen
### pero deja menos comisión?

El P2P es gratis: mucho volumen,
cero ingresos. Hay que detectarlo.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/paypal_pagos.csv'
)

r = (
  df
  .groupby('tipo')
  .agg(
    volumen=('volumen_usd', 'sum'),
    comision=('comision_usd', 'sum'),
  )
)

r['take_rate_%'] = (
  r['comision'] / r['volumen'] * 100
).round(2)
r = r.round(0).sort_values(
  'volumen', ascending=False)

print('Volumen vs comisión:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuál es el take rate
### medio global de
### la plataforma?

Take rate = comisión / volumen. Es
EL número que mira Wall Street.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/paypal_pagos.csv'
)

vol = df['volumen_usd'].sum()
com = df['comision_usd'].sum()
take = com / vol * 100

print(f'Volumen total: '
      f'{vol:,.0f} USD')
print(f'Comisión total: '
      f'{com:,.0f} USD')
print(f'Take rate: {take:.2f}%')
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Las remesas
### internacionales valen
### más la pena? (ticket)

Comparamos ticket medio de pagos
internacionales vs nacionales.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/paypal_pagos.csv'
)

df['tipo_env'] = df['internacional'].map(
  {1: 'Internacional', 0: 'Nacional'}
)

r = (
  df
  .groupby('tipo_env')
  .agg(
    ticket=('volumen_usd', 'mean'),
    com_media=('comision_usd',
               'mean'),
    n=('pago_id', 'count'),
  )
  .round(2)
)

print('Internacional vs nacional:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Desde qué dispositivo
### se mueve más
### dinero? (móvil manda)

Si el móvil domina, el producto hay
que diseñarlo mobile-first.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/paypal_pagos.csv'
)

total = df['volumen_usd'].sum()

r = (
  df
  .groupby('dispositivo')
  ['volumen_usd']
  .sum()
  .sort_values(ascending=False)
)

tabla = pd.DataFrame({
  'volumen_M': (r / 1e3).round(1),
  '%': (r / total * 100).round(1),
})

print('Volumen por dispositivo:')
print(tabla)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué divisas aportan
### el 80% del volumen?
### (regla de Pareto)

El % acumulado nos dice en qué pocas
divisas concentramos el negocio.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/paypal_pagos.csv'
)

r = (
  df
  .groupby('divisa')['volumen_usd']
  .sum()
  .sort_values(ascending=False)
)

acum = (
  r.cumsum() / r.sum() * 100
).round(1)

tabla = pd.DataFrame({
  'volumen': r.round(0),
  '%_acum': acum,
})

print('Pareto de divisas:')
print(tabla)
```
