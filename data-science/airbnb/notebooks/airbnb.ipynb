### 💼 PREGUNTA DE ENTREVISTA:
### ¿Cuál es el barrio más caro de Nueva York para alojarse en Airbnb?

```python
import pandas as pd

df = pd.read_csv('../datasets/airbnb_nyc.csv')

result = df.groupby('neighbourhood_group')['price'].mean().sort_values(ascending=False)
print(f"Barrio más caro: {result.idxmax()} (${result.max():.0f}/noche de media)")
print()
print(result.round(0))
```

### 💼 PREGUNTA DE ENTREVISTA:
### ¿Qué tipo de alojamiento es más común y cuál es más caro?

```python
import pandas as pd

df = pd.read_csv('../datasets/airbnb_nyc.csv')

count = df['room_type'].value_counts()
avg_price = df.groupby('room_type')['price'].mean().sort_values(ascending=False)

result = pd.DataFrame({'cantidad': count, 'precio_medio': avg_price.round(0)})
print(f"Más común: {count.idxmax()} ({count.max():,} alojamientos)")
print(f"Más caro: {avg_price.idxmax()} (${avg_price.max():.0f}/noche)")
print()
print(result)
```

### 💼 PREGUNTA DE ENTREVISTA:
### ¿Quiénes son los hosts con más propiedades y cuánto ganan?

```python
import pandas as pd

df = pd.read_csv('../datasets/airbnb_nyc.csv')

# Contar propiedades por host y calcular ingreso estimado
host_stats = df.groupby('host_name').agg(
    propiedades=('id', 'count'),
    precio_medio=('price', 'mean'),
    reviews_total=('number_of_reviews', 'sum')
).sort_values('propiedades', ascending=False).head(10)

host_stats['ingreso_estimado_mes'] = (host_stats['precio_medio'] * host_stats['reviews_total'] / 12).round(0)

print("🏆 TOP 10 HOSTS CON MÁS PROPIEDADES EN NYC")
print(host_stats[['propiedades', 'precio_medio', 'ingreso_estimado_mes']].to_string())
```

### 💼 PREGUNTA DE ENTREVISTA:
### ¿Cuántos días de media está disponible un Airbnb al año por zona?

```python
import pandas as pd

df = pd.read_csv('../datasets/airbnb_nyc.csv')

result = df.groupby('neighbourhood_group')['availability_365'].mean().sort_values(ascending=False)

print("📅 DISPONIBILIDAD MEDIA (días/año) POR ZONA")
print(result.round(0).to_string())
print(f"\nZona con más disponibilidad: {result.idxmax()} ({result.max():.0f} días/año)")
```

### 💼 PREGUNTA DE ENTREVISTA:
### ¿Cuál es la relación entre precio y número de reviews?

```python
import pandas as pd

df = pd.read_csv('../datasets/airbnb_nyc.csv')

# Filtrar outliers (precio < 1000)
df_clean = df[df['price'] < 1000]

correlation = df_clean['price'].corr(df_clean['number_of_reviews'])
print(f"Correlación precio vs reviews: {correlation:.3f}")
print(f"\n{'📉 Correlación NEGATIVA' if correlation < 0 else '📈 Correlación POSITIVA'}")
print("Los alojamientos más baratos tienden a tener más reviews (se reservan más).")
```
