### 💼 PREGUNTA ENTREVISTA:
### ¿Cuál es el Airbnb
### más caro de Barcelona?

```python
import pandas as pd 

df = pd.read_csv(
    '../datasets/airbnb_barcelona.csv'
)

cols = [
    'name',
    'neighbourhood',
    'price'
]

top = (df.nlargest(5, 'price')[cols])

print(top.to_string(index=False))
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuántos pisos llevan
### +1 año sin actividad?

```python
import pandas as pd

df = pd.read_csv(
    '../datasets/airbnb_madrid.csv'
)

df['last_review'] = pd.to_datetime(
    df['last_review']
)

hoy = pd.Timestamp.today()

df['dias'] = (
    (hoy - df['last_review']).dt.days
)

activos = df[df['last_review'].notna()]
viejos = activos[activos['dias']>365]

pct = len(viejos)/len(activos)*100

print(f"{len(viejos):,} pisos")
print(f"con +1 año sin review")
print(f"({pct:.1f}% del total)")
```
