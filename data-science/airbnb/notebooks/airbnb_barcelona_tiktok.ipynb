### 💼 PREGUNTA ENTREVISTA:
### ¿Cuántas noches mínimas
### piden por barrio en BCN?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/airbnb_barcelona.csv'
)

result = (
  df
  .groupby('neighbourhood')
  ['minimum_nights']
  .mean()
  .sort_values(ascending=False)
  .head(10)
)

print(result.round(1))
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuántos alojamientos
### tienen 0 reviews?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/airbnb_barcelona.csv'
)

zero = df[
  df['number_of_reviews'] == 0
]

pct = len(zero) / len(df) * 100

print(f"{len(zero):,} sin reviews")
print(f"({pct:.1f}% del total)")
print()
print("Precio medio:")
print(f"  Sin reviews: "
      f"{zero['price'].mean():.0f}€")
print(f"  Con reviews: "
      f"{df[df['number_of_reviews']>0]['price'].mean():.0f}€")
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Quién tiene un monopolio
### de pisos en Barcelona?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/airbnb_barcelona.csv'
)

big = df[
  df['calculated_host_listings_count']
  > 10
]

top = (
  big
  .groupby('host_name')
  ['id'].count()
  .sort_values(ascending=False)
  .head(10)
)

print(f"{len(big):,} pisos en")
print(f"manos de mega-hosts")
print(f"({len(big)/len(df)*100:.1f}%)")
print()
print(top)
```
