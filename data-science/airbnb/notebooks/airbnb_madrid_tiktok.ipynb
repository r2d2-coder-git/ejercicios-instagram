### 💼 PREGUNTA ENTREVISTA:
### ¿Cuál es el distrito
### más caro de Madrid?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/airbnb_madrid.csv'
)

result = (
  df
  .groupby('neighbourhood_group')
  ['price']
  .median()
  .sort_values(ascending=False)
)

print("€/noche (mediana):")
print(result)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué barrios tienen
### +500 alojamientos?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/airbnb_madrid.csv'
)

counts = (
  df['neighbourhood']
  .value_counts()
)

big = counts[counts > 500]

pct = big.sum() / len(df) * 100

print(f"{len(big)} barrios con +500:")
print(big)
print(f"\n= {pct:.0f}% de Madrid")
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Es más caro un piso
### CON licencia o SIN?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/airbnb_madrid.csv'
)

df['licencia'] = (
  df['license'].notna()
)

result = (
  df
  .groupby('licencia')
  ['price']
  .median()
)

result.index = [
  'Sin licencia',
  'Con licencia'
]

print("Precio mediano €/noche:")
print(result)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuántos pisos llevan
### +1 año sin actividad? (TODO)

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/airbnb_madrid.csv'
)

df['last_review'] = pd.to_datetime(
  df['last_review']
)

hoy = pd.Timestamp.today()

df['dias'] = (
  (hoy - df['last_review']).dt.days
)

activos = df[df['last_review'].notna()]
viejos = activos[activos['dias'] > 365]

pct = len(viejos)/len(activos)*100

print(f"{len(viejos):,} pisos")
print(f"con +1 año sin review")
print(f"({pct:.1f}% del total)")
print()
print("Probablemente inactivos.")
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuánto ganarías al mes
### con un Airbnb en Madrid?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/airbnb_madrid.csv'
)

# reviews_per_month × 3 = noches
df['ingreso_mes'] = (
  df['price']
  * df['reviews_per_month']
  * 3
)

result = (
  df
  .groupby('neighbourhood_group')
  ['ingreso_mes']
  .median()
  .sort_values(ascending=False)
)

print("€/mes estimado (mediana):")
print(result.round(0))
```
