### 💼 PREGUNTA ENTREVISTA:
### ¿Qué categoría tiene
### mejor engagement?
### (save rate real)

Save rate = guardados / impresiones.
Es la métrica estrella de Pinterest.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/pinterest_pins.csv'
)

g = (
  df
  .groupby('categoria')
  [['guardados', 'impresiones']]
  .sum()
)

g['save_rate_%'] = (
  g['guardados']
  / g['impresiones'] * 100
).round(2)

print('Save rate por categoría:')
print(
  g['save_rate_%']
  .sort_values(ascending=False)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿El vídeo rinde más
### que la imagen fija
### en clics (CTR)?

CTR = clics / impresiones. ¿Merece
la pena priorizar vídeo en el feed?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/pinterest_pins.csv'
)

g = (
  df
  .groupby('es_video')
  [['clics', 'impresiones']]
  .sum()
)

g['ctr_%'] = (
  g['clics']
  / g['impresiones'] * 100
).round(3)

g.index = g.index.map(
  {True: 'Vídeo', False: 'Imagen'})

print('CTR por formato:')
print(g['ctr_%'])
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué categorías mueven
### el 80% de los
### guardados? (Pareto)

Regla 80/20: pocas categorías suelen
concentrar casi todo el valor.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/pinterest_pins.csv'
)

s = (
  df
  .groupby('categoria')
  ['guardados']
  .sum()
  .sort_values(ascending=False)
)

acum = (
  s.cumsum() / s.sum() * 100
).round(1)

tabla = pd.DataFrame({
  'guardados': s,
  '%_acum': acum,
})

print('Pareto de guardados:')
print(tabla)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Hay estacionalidad?
### ¿Qué meses explotan
### en guardados?

Pinterest vive de planificar (bodas,
navidad...). Buscamos los picos.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/pinterest_pins.csv'
)

s = (
  df
  .groupby('mes')['guardados']
  .sum()
)

media = s.mean()
idx = (s / media * 100).round(0)

tabla = pd.DataFrame({
  'guardados': s,
  'idx_vs_media': idx,
}).sort_values('guardados',
  ascending=False)

print('Guardados por mes:')
print(tabla.head(6))
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Clicar y guardar van
### de la mano o son
### cosas distintas?

Correlación entre CTR y save rate
a nivel pin. Si es baja, miden intenciones
diferentes.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/pinterest_pins.csv'
)

d = df[df['impresiones'] > 0].copy()
d['ctr'] = (
  d['clics'] / d['impresiones']
)
d['save'] = (
  d['guardados'] / d['impresiones']
)

corr = d['ctr'].corr(d['save'])

print('Correlación CTR vs save rate:')
print(round(corr, 3))
if abs(corr) < 0.3:
    print('-> Miden cosas distintas')
else:
    print('-> Van bastante juntas')
```
