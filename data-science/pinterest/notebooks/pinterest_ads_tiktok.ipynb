### 💼 PREGUNTA ENTREVISTA:
### ¿Qué formato de anuncio
### convierte más barato?
### (coste por conversión)

CPA = inversión / conversiones.
Es el euro que de verdad importa.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/pinterest_ads.csv'
)

g = (
  df
  .groupby('formato')
  [['inversion_usd',
    'conversiones']]
  .sum()
)

g['cpa_usd'] = (
  g['inversion_usd']
  / g['conversiones']
).round(2)

print('CPA por formato:')
print(
  g['cpa_usd']
  .sort_values()
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué sector tiene el
### embudo más eficiente
### de clic a venta?

CTR y CVR juntos. Dónde el usuario
hace clic y además compra.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/pinterest_ads.csv'
)

g = (
  df
  .groupby('sector')
  [['impresiones', 'clics',
    'conversiones']]
  .sum()
)

g['ctr_%'] = (
  g['clics']
  / g['impresiones'] * 100
).round(2)

g['cvr_%'] = (
  g['conversiones']
  / g['clics'] * 100
).round(2)

print('Embudo por sector:')
print(
  g[['ctr_%', 'cvr_%']]
  .sort_values('cvr_%',
    ascending=False)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿El 20% de campañas
### se lleva el 80% de
### la inversión? (Pareto)

Regla de Pareto sobre el gasto
publicitario. ¿Está concentrado?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/pinterest_ads.csv'
)

s = (
  df['inversion_usd']
  .sort_values(ascending=False)
  .reset_index(drop=True)
)

top20 = int(len(s) * 0.2)
pct = (
  s.head(top20).sum()
  / s.sum() * 100
)

print(f'Campañas: {len(s)}')
print(f'Top 20%: {top20}')
print(
  f'Se llevan: {pct:.1f}% '
  'del gasto'
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿En qué tramo de CPC
### están las campañas
### más rentables?

Agrupamos el CPC en tramos y vemos
la conversión media de cada uno.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/pinterest_ads.csv'
)

df['tramo_cpc'] = pd.cut(
  df['cpc_usd'],
  bins=[0, .3, .6, 1, 99],
  labels=['bajo', 'medio',
          'alto', 'premium'],
)

r = (
  df
  .groupby('tramo_cpc',
    observed=True)
  .agg(
    conv=('conversiones',
          'mean'),
    inv=('inversion_usd',
         'mean'),
    n=('cpc_usd', 'size'),
  )
  .round(2)
)

print('Conversión por tramo CPC:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué combo sector +
### formato da el mejor
### retorno por conversión?

Cruce de dos dimensiones. El ROI
fino está en el detalle.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/pinterest_ads.csv'
)

g = (
  df
  .groupby(['sector',
            'formato'])
  [['inversion_usd',
    'conversiones']]
  .sum()
)

g = g[g['conversiones'] > 0]
g['cpa'] = (
  g['inversion_usd']
  / g['conversiones']
).round(2)

print('Mejores combos (CPA bajo):')
print(
  g['cpa']
  .sort_values()
  .head(8)
)
```
