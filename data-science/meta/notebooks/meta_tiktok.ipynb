### 💼 PREGUNTA ENTREVISTA:
### ¿Qué plataforma del
### grupo tiene más usuarios
### activos al mes?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/meta_usuarios.csv'
)

# Último mes disponible
ult = df['mes'].max()
foto = df[df['mes'] == ult]

r = (
  foto
  .groupby('plataforma')
  ['mau_millones']
  .sum()
  .sort_values(ascending=False)
  .round(0)
)

print(f'MAU por plataforma ({ult}):')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué países nos dan
### más dinero aunque no
### sean los más grandes?

El ARPU (ingreso por usuario) manda:
pocos usuarios ricos > muchos pobres.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/meta_usuarios.csv'
)

r = (
  df
  .groupby('pais')
  .agg(
    ingresos=('ingresos_pub_musd',
              'sum'),
    arpu=('arpu_usd', 'mean'),
    mau=('mau_millones', 'mean'),
  )
  .round(1)
  .sort_values('ingresos',
    ascending=False)
  .head(8)
)

print('Top países por ingresos:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué plataforma engancha
### más minutos al día
### a cada usuario?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/meta_usuarios.csv'
)

r = (
  df
  .groupby('plataforma')
  .agg(
    minutos=('minutos_diarios',
             'mean'),
    arpu=('arpu_usd', 'mean'),
  )
  .round(1)
  .sort_values('minutos',
    ascending=False)
)

print('Engagement por plataforma:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuánto creció Threads
### en el año? (primer mes
### vs último)

Threads es la apuesta nueva.
¿Está despegando de verdad?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/meta_usuarios.csv'
)

th = (
  df[df['plataforma'] == 'Threads']
  .groupby('mes')['mau_millones']
  .sum()
  .sort_index()
)

ini = th.iloc[0]
fin = th.iloc[-1]
crec = (fin / ini - 1) * 100

print(f'Threads inicio: {ini:.0f}M')
print(f'Threads final:  {fin:.0f}M')
print(f'Crecimiento: +{crec:.1f}%')
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué plataforma retiene
### mejor? (DAU/MAU, el
### 'stickiness')

DAU/MAU dice cuántos días al mes
entra el usuario. Es LA métrica.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/meta_usuarios.csv'
)

g = (
  df
  .groupby('plataforma')
  [['dau_millones', 'mau_millones']]
  .sum()
)

g['stickiness_%'] = (
  g['dau_millones']
  / g['mau_millones'] * 100
).round(1)

print(
  g['stickiness_%']
  .sort_values(ascending=False)
)
```
