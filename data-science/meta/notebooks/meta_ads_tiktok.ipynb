### 💼 PREGUNTA ENTREVISTA:
### ¿Qué plataforma da
### el CPA más bajo?
### (coste por conversión)

CPA = inversión / conversiones. Es
la métrica que mira el anunciante.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/meta_ads.csv'
)

g = (
  df
  .groupby('plataforma')
  .agg(
    inv=('inversion_usd', 'sum'),
    conv=('conversiones', 'sum'),
  )
)

g['cpa_usd'] = (
  g['inv'] / g['conv']
).round(2)

print('CPA por plataforma:')
print(
  g.sort_values('cpa_usd')
  .round(0)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué sector convierte
### mejor y cuál solo
### quema presupuesto?

Tasa conversión = conversiones /
clics. Separa el grano de la paja.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/meta_ads.csv'
)

g = (
  df
  .groupby('sector')
  .agg(
    clics=('clics', 'sum'),
    conv=('conversiones', 'sum'),
    inv=('inversion_usd', 'sum'),
  )
)

g['conv_pct'] = (
  g['conv'] / g['clics'] * 100
).round(2)

print('Conversión por sector:')
print(
  g.sort_values('conv_pct',
    ascending=False)
  .round(0)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Más inversión =
### mejor CTR? (¿o se
### satura?)

Troceamos inversión en tramos y
miramos el CTR medio de cada uno.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/meta_ads.csv'
)

df['tramo'] = pd.qcut(
  df['inversion_usd'], 4,
  labels=['Q1', 'Q2', 'Q3', 'Q4'],
)

r = (
  df
  .groupby('tramo', observed=True)
  .agg(
    ctr=('ctr_pct', 'mean'),
    cpc=('cpc_usd', 'mean'),
  )
  .round(2)
)

print('CTR/CPC por tramo de gasto:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿El objetivo de campaña
### cambia el coste por
### conversión?

Conversiones vs Reconocimiento...
cada objetivo paga un precio distinto.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/meta_ads.csv'
)

g = (
  df
  .groupby('objetivo')
  .agg(
    inv=('inversion_usd', 'sum'),
    conv=('conversiones', 'sum'),
    n=('campana_id', 'count'),
  )
)

g['cpa_usd'] = (
  g['inv'] / g['conv']
).round(2)

print('CPA por objetivo:')
print(
  g.sort_values('cpa_usd')
  .round(0)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿El CTR va ligado al
### CPC? (correlación
### que paga el cliente)

Un anuncio con buen CTR, ¿se
abarata? Medimos la correlación.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/meta_ads.csv'
)

d = df[
  (df['ctr_pct'] > 0)
  & (df['cpc_usd'] > 0)
]

corr = (
  d['ctr_pct']
  .corr(d['cpc_usd'])
)

print('Correlación CTR vs CPC:')
print(round(corr, 3))

signo = (
  'inversa' if corr < 0
  else 'directa'
)
print(f'Relación {signo}')
```
