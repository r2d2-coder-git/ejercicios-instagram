### 💼 PREGUNTA ENTREVISTA:
### ¿Qué modelo nos
### genera más gasto
### total en la factura?

El coste manda al elegir modelo.
¿Dónde se va el dinero?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/openai_uso_api.csv'
)

r = (
  df
  .groupby('modelo')['coste_usd']
  .sum()
  .sort_values(ascending=False)
)

pct = (r / r.sum() * 100).round(1)

tabla = pd.DataFrame({
  'coste_usd': r.round(2),
  '%': pct,
})

print('Gasto por modelo:')
print(tabla)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuánto cuesta de
### media 1.000 tokens
### en cada modelo?

Coste normalizado por 1k tokens para
comparar modelos de forma justa.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/openai_uso_api.csv'
)

df['tokens'] = (
  df['tokens_entrada']
  + df['tokens_salida']
)

g = (
  df
  .groupby('modelo')
  [['coste_usd', 'tokens']]
  .sum()
)

g['usd_1k_tok'] = (
  g['coste_usd']
  / g['tokens'] * 1000
).round(4)

print('Coste por 1k tokens:')
print(
  g['usd_1k_tok']
  .sort_values(ascending=False)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Hay trade-off entre
### coste y latencia
### al elegir modelo?

Lo barato suele ser rápido. Medimos
coste medio vs latencia media.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/openai_uso_api.csv'
)

r = (
  df
  .groupby('modelo')
  .agg(
    coste=('coste_usd', 'mean'),
    latencia=('latencia_s',
              'mean'),
  )
  .round(4)
  .sort_values('latencia')
)

print('Coste vs latencia:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué tipo de petición
### consume más tokens
### de salida?

Los tokens de salida son los caros.
¿Chat, código, visión...?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/openai_uso_api.csv'
)

r = (
  df
  .groupby('tipo_peticion')
  .agg(
    tok_salida=('tokens_salida',
                'mean'),
    coste=('coste_usd', 'mean'),
    n=('coste_usd', 'size'),
  )
  .round(2)
  .sort_values('tok_salida',
    ascending=False)
)

print('Por tipo de petición:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué segmento de
### cliente deja más
### dinero y a qué ritmo?

Gasto total y gasto medio por
petición según tipo de cliente.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/openai_uso_api.csv'
)

r = (
  df
  .groupby('cliente')
  .agg(
    total=('coste_usd', 'sum'),
    medio=('coste_usd', 'mean'),
    peticiones=('coste_usd',
                'size'),
  )
  .round(3)
  .sort_values('total',
    ascending=False)
)

print('Gasto por tipo de cliente:')
print(r)
```
