### 💼 PREGUNTA ENTREVISTA:
### ¿Qué puesto paga más
### y cuánto sube de
### Junior a Senior?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/linkedin_empleos.csv'
)

t = (
  df
  .pivot_table(
    index='puesto',
    columns='nivel',
    values='salario_eur',
    aggfunc='median',
  )
  [['Junior', 'Senior']]
)

t['salto_%'] = (
  t['Senior'] / t['Junior'] - 1
) * 100

print('Mediana salario (EUR):')
print(
  t.sort_values('Senior',
    ascending=False).round(0)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué skill técnica
### se paga mejor
### en el mercado?

Mediana por skill principal. Ojo:
no es lo más pedido, es lo mejor pagado.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/linkedin_empleos.csv'
)

r = (
  df
  .groupby('skill_principal')
  .agg(
    salario=('salario_eur',
             'median'),
    ofertas=('salario_eur',
             'size'),
  )
  .round(0)
  .sort_values('salario',
    ascending=False)
)

print('Salario mediano por skill:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿El trabajo remoto
### paga más o menos
### que el presencial?

Comparamos mediana por modalidad
a mismo nivel para que sea justo.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/linkedin_empleos.csv'
)

r = (
  df[df['nivel'] == 'Senior']
  .groupby('modalidad')
  ['salario_eur']
  .median()
  .round(0)
  .sort_values(ascending=False)
)

base = r['Presencial']
dif = (r / base - 1) * 100

print('Senior - mediana por modalidad:')
print(r)
print('\nvs presencial (%):')
print(dif.round(1))
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Dónde hay menos
### competencia por oferta?
### (candidatos/vacante)

Pocos candidatos por oferta = más
fácil entrar. Buscamos el hueco.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/linkedin_empleos.csv'
)

r = (
  df
  .groupby('puesto')
  .agg(
    cand_medio=('candidatos',
                'mean'),
    salario=('salario_eur',
             'median'),
  )
  .round(0)
  .sort_values('cand_medio')
)

print('Menos competencia arriba:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué sectores pagan
### por encima de la
### media del mercado?

Índice 100 = media global. >100
significa que paga mejor que la media.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/linkedin_empleos.csv'
)

media = df['salario_eur'].mean()

r = (
  df
  .groupby('sector')
  ['salario_eur']
  .mean()
)

indice = (r / media * 100).round(1)
indice = indice.sort_values(
  ascending=False)

print('Índice salarial (100=media):')
print(indice)
```
