### 💼 PREGUNTA ENTREVISTA:
### ¿Más conexiones
### traen más visitas
### al perfil? (corr)

La red importa, pero ¿cuánto? Vemos
correlación y visitas por tramo.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/linkedin_perfiles.csv'
)

corr = (
  df['conexiones']
  .corr(df['visualizaciones_perfil'])
)

df['tramo'] = pd.qcut(
  df['conexiones'],
  q=4,
  labels=['Q1', 'Q2', 'Q3', 'Q4'],
)

r = (
  df
  .groupby('tramo', observed=True)
  ['visualizaciones_perfil']
  .mean()
  .round(0)
)

print(f'Correlación: {corr:.2f}')
print('\nVisitas por tramo red:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Los usuarios premium
### tienen más visitas
### al perfil? (lift)

Comparamos visitas medias premium
vs gratis a igualdad de experiencia.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/linkedin_perfiles.csv'
)

r = (
  df
  .groupby('premium')
  ['visualizaciones_perfil']
  .mean()
)

lift = (r[1] / r[0] - 1) * 100

print(f'Gratis: {r[0]:,.0f}')
print(f'Premium: {r[1]:,.0f}')
print(f'Lift premium: +{lift:.1f}%')
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué sector tiene
### los perfiles más
### activos? (posts)

Publicaciones por años de experiencia
mide la actividad real por sector.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/linkedin_perfiles.csv'
)

df['posts_ano'] = (
  df['publicaciones']
  / df['anios_exp'].clip(lower=1)
)

r = (
  df
  .groupby('sector')
  .agg(
    posts=('publicaciones',
           'mean'),
    ritmo=('posts_ano', 'mean'),
  )
  .round(2)
  .sort_values('ritmo',
    ascending=False)
)

print('Actividad por sector:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué sector convierte
### mejor a premium?
### (tasa)

Tasa de premium por sector guía
dónde está el cliente que paga.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/linkedin_perfiles.csv'
)

r = (
  df
  .groupby('sector')
  .agg(
    premium=('premium', 'mean'),
    users=('premium', 'size'),
  )
)

r['premium_%'] = (
  r['premium'] * 100
).round(1)
r = r.drop(columns='premium')

print('Conversión premium x sector:')
print(
  r.sort_values('premium_%',
    ascending=False)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Los superconectados
### (top 10%) acaparan
### las visitas?

Pareto de la red: ¿el 10% con más
contactos concentra la visibilidad?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/linkedin_perfiles.csv'
)

df['grupo'] = pd.qcut(
  df['conexiones'],
  q=[0, 0.9, 1.0],
  labels=['90% base',
          'Top 10%'],
)

r = (
  df
  .groupby('grupo', observed=True)
  ['visualizaciones_perfil']
  .sum()
)

pct = (r / r.sum() * 100).round(1)

print('Visitas totales por grupo:')
print(r)
print('\n% de visitas:')
print(pct)
```
