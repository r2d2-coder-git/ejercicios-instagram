### 💼 PREGUNTA ENTREVISTA:
### ¿Qué caso de uso
### dispara los rechazos
### de seguridad?

El rechazo de seguridad es clave en
Anthropic. ¿Dónde se concentra?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/anthropic_claude.csv'
)

r = (
  df
  .groupby('caso_uso')
  ['rechazo_seguridad']
  .mean()
  .mul(100)
  .round(2)
  .sort_values(ascending=False)
)

print('% rechazos por caso de uso:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Un rechazo de
### seguridad hunde la
### satisfacción?

Comparamos satisfacción media con
y sin rechazo. Mide la fricción.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/anthropic_claude.csv'
)

r = (
  df
  .groupby('rechazo_seguridad')
  ['satisfaccion']
  .mean()
  .round(2)
)

r.index = r.index.map(
  {True: 'Con rechazo',
   False: 'Sin rechazo'})

caida = r['Con rechazo'] \
  - r['Sin rechazo']

print('Satisfacción media:')
print(r)
print(f'\nCaída: {caida:.2f} puntos')
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué modelo de Claude
### da mejor satisfacción
### por dólar gastado?

Satisfacción alta y coste bajo es el
equilibrio ideal para el cliente.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/anthropic_claude.csv'
)

r = (
  df
  .groupby('modelo')
  .agg(
    satisf=('satisfaccion',
            'mean'),
    coste=('coste_usd', 'mean'),
  )
)

r['satisf_x_usd'] = (
  r['satisf'] / r['coste']
).round(1)

print('Satisfacción por dólar:')
print(
  r.round(3)
  .sort_values('satisf_x_usd',
    ascending=False)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué segmento usa
### Claude de forma más
### intensiva (tokens)?

Tokens totales medios por conversación
según el tipo de cliente.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/anthropic_claude.csv'
)

df['tokens'] = (
  df['tokens_entrada']
  + df['tokens_salida']
)

r = (
  df
  .groupby('cliente')
  .agg(
    tok_medio=('tokens', 'mean'),
    coste=('coste_usd', 'sum'),
    convs=('tokens', 'size'),
  )
  .round(2)
  .sort_values('tok_medio',
    ascending=False)
)

print('Intensidad por cliente:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué casos de uso
### son el core del
### negocio? (Pareto)

Volumen de conversaciones por caso y
su % acumulado sobre el total.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/anthropic_claude.csv'
)

s = (
  df['caso_uso']
  .value_counts()
)

acum = (
  s.cumsum() / s.sum() * 100
).round(1)

tabla = pd.DataFrame({
  'convs': s,
  '%_acum': acum,
})

print('Pareto de casos de uso:')
print(tabla)
```
