### 💼 PREGUNTA ENTREVISTA:
### ¿Qué departamento
### factura más y cuál
### deja más margen €?

Alimentación vende mucho pero con
poco margen. El margen € reequilibra.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/walmart_ventas.csv'
)

df['margen_usd'] = (
  df['ventas_usd'] * df['margen']
)

r = (
  df
  .groupby('departamento')
  .agg(
    ventas=('ventas_usd', 'sum'),
    margen=('margen_usd', 'sum'),
  )
  .round(0)
  .sort_values('ventas',
    ascending=False)
)

print('Ventas y margen por depto:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuánto suben las
### ventas en semanas
### festivas? (lift)

Black Friday y Navidad: comparamos
ventas festivas vs normales.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/walmart_ventas.csv'
)

r = (
  df
  .groupby('festivo')['ventas_usd']
  .mean()
)

lift = (r[1] / r[0] - 1) * 100

print(f'Media normal: {r[0]:,.0f}')
print(f'Media festivo: {r[1]:,.0f}')
print(f'Lift festivo: +{lift:.1f}%')
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué 5 tiendas
### venden más por
### semana? (ranking)

El ranking de tiendas top guía dónde
replicar buenas prácticas.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/walmart_ventas.csv'
)

r = (
  df
  .groupby('tienda')['ventas_usd']
  .mean()
  .round(0)
  .sort_values(ascending=False)
  .head(5)
)

print('Top 5 tiendas (venta/sem):')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué región vende
### más y qué formato
### de tienda gana?

Cruzamos región x tipo de tienda
para ver dónde expandir.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/walmart_ventas.csv'
)

r = pd.pivot_table(
  df,
  index='region',
  columns='tipo_tienda',
  values='ventas_usd',
  aggfunc='sum',
)

r = (r / 1e6).round(1)

print('Ventas (M$) región x tipo:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué deptos venden
### barato por unidad?
### (precio medio)

Precio medio = ventas / unidades.
Define el posicionamiento del depto.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/walmart_ventas.csv'
)

g = (
  df
  .groupby('departamento')
  .agg(
    ventas=('ventas_usd', 'sum'),
    unid=('unidades', 'sum'),
  )
)

g['precio_medio'] = (
  g['ventas'] / g['unid']
).round(2)

print('Precio medio por unidad:')
print(
  g['precio_medio']
  .sort_values()
)
```
