### 💼 PREGUNTA ENTREVISTA:
### ¿Qué categoría de
### comercio mueve más
### gasto y qué ticket?

El ticket medio distingue el gasto
del día a día del gasto grande.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/visa_transacciones.csv'
)

ok = df[df['aprobada'] == 1]

r = (
  ok
  .groupby('categoria_comercio')
  .agg(
    gasto=('importe_usd', 'sum'),
    ticket=('importe_usd', 'mean'),
    n=('tx_id', 'count'),
  )
  .round(1)
  .sort_values('gasto',
    ascending=False)
)

print('Gasto por categoría:')
print(r.head(6))
```

### 💼 PREGUNTA ENTREVISTA:
### ¿A qué horas se
### concentra el gasto?
### (franjas del día)

Agrupamos las 24h en franjas para
ver cuándo compra la gente.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/visa_transacciones.csv'
)

df['franja'] = pd.cut(
  df['hora'],
  bins=[-1, 5, 11, 17, 23],
  labels=['Madrugada', 'Mañana',
          'Tarde', 'Noche'],
)

r = (
  df
  .groupby('franja', observed=True)
  ['importe_usd']
  .agg(['sum', 'count'])
  .round(0)
)

print('Gasto por franja horaria:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué canal rechaza
### más? (online vs
### físico vs contactless)

Tasa de rechazo = (1-aprobadas).
Online suele tener más fricción.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/visa_transacciones.csv'
)

r = (
  df
  .groupby('canal')
  .agg(
    rechazo_pct=('aprobada',
      lambda s: (1 - s.mean()) * 100),
    n=('tx_id', 'count'),
  )
  .round(2)
  .sort_values('rechazo_pct',
    ascending=False)
)

print('Rechazo por canal:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Crédito o débito
### mueven el gasto
### grande? (tabla cruzada)

Cruzamos tipo de tarjeta con
categoría para ver quién paga qué.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/visa_transacciones.csv'
)

r = pd.pivot_table(
  df,
  index='categoria_comercio',
  columns='tipo_tarjeta',
  values='importe_usd',
  aggfunc='mean',
).round(0)

print('Ticket medio por tipo:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué países concentran
### el gasto por
### transacción más alto?

Ordenamos países por ticket medio
para priorizar mercados premium.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/visa_transacciones.csv'
)

r = (
  df[df['aprobada'] == 1]
  .groupby('pais')
  .agg(
    ticket=('importe_usd', 'mean'),
    gasto=('importe_usd', 'sum'),
  )
  .round(1)
  .sort_values('ticket',
    ascending=False)
)

print('Ticket medio por país:')
print(r)
```
