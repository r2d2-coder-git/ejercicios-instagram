### 💼 PREGUNTA ENTREVISTA:
### ¿Qué GPU da más
### rendimiento por dólar?
### (el verdadero chollo)

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/nvidia_gpus.csv'
)

g = (
  df[df['segmento'] == 'Gaming']
  .groupby('modelo')
  .agg(
    precio=('precio_usd', 'mean'),
    perf=('perf_score', 'first'),
  )
)

g['perf_por_dolar'] = (
  g['perf'] / g['precio'] * 100
).round(2)

print('Top rendimiento/precio:')
print(
  g.sort_values('perf_por_dolar',
    ascending=False)
  .head(5)
  .round(1)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuánto sobreprecio
### sobre el MSRP paga
### el cliente de verdad?

MSRP = precio recomendado. Lo que
pagas en tienda suele ser más.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/nvidia_gpus.csv'
)

df['sobreprecio_%'] = (
  df['precio_usd']
  / df['msrp_usd'] - 1
) * 100

r = (
  df
  .groupby('segmento')
  ['sobreprecio_%']
  .mean()
  .round(1)
  .sort_values(ascending=False)
)

print('Sobreprecio medio vs MSRP:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿El negocio de datacenter
### (IA) pesa más que el
### gaming? (facturación)

El boom de la IA cambió a Nvidia.
Facturación = precio x stock vendido.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/nvidia_gpus.csv'
)

df['facturacion'] = (
  df['precio_usd'] * df['stock']
)

r = (
  df
  .groupby('segmento')
  ['facturacion']
  .sum()
)

r = (r / 1e6).round(1)
pct = (r / r.sum() * 100).round(1)

tabla = pd.DataFrame({
  'factur_M': r,
  '%': pct,
}).sort_values('%', ascending=False)

print('Facturación por segmento:')
print(tabla)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cada nueva arquitectura
### mejora el rendimiento
### por vatio?

Eficiencia = perf / consumo. Es
clave en datacenters (factura de luz).

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/nvidia_gpus.csv'
)

gpus = df.drop_duplicates('modelo')

gpus = gpus.assign(
  perf_por_watt=(
    gpus['perf_score']
    / gpus['tdp_w']
  )
)

r = (
  gpus
  .groupby('arquitectura')
  .agg(
    perf_watt=('perf_por_watt',
               'mean'),
    anio=('anio_lanzamiento',
          'first'),
  )
  .round(3)
  .sort_values('anio')
)

print('Eficiencia por arquitectura:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuándo tocó fondo el
### precio de la RTX 4090?
### (mejor momento de compra)

Buscamos el mes más barato de la
estrella gaming para avisar al cliente.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/nvidia_gpus.csv'
)

s = (
  df[df['modelo']
     == 'GeForce RTX 4090']
  .groupby('fecha')['precio_usd']
  .mean()
  .round(0)
)

minimo = s.idxmin()
maximo = s.idxmax()

print(f'Más barata: {minimo} '
      f'({s.min():.0f} USD)')
print(f'Más cara:   {maximo} '
      f'({s.max():.0f} USD)')
print(f'Ahorro: '
      f'{s.max()-s.min():.0f} USD')
```
