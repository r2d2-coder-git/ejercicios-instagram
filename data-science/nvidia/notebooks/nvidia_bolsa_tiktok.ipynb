### 💼 PREGUNTA ENTREVISTA:
### ¿Cuánto se revalorizó
### la acción en todo
### el periodo?

Comparamos primer y último precio.
El efecto del boom de la IA.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/nvidia_bolsa.csv'
)

ini = df['precio_accion'].iloc[0]
fin = df['precio_accion'].iloc[-1]

ret = (fin / ini - 1) * 100

print(f'Precio inicial: {ini} USD')
print(f'Precio final:   {fin} USD')
print(f'Revalorización: '
      f'{ret:.0f}%')
print(f'Multiplicó x'
      f'{fin/ini:.1f}')
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué mes fue el más
### volátil? (riesgo
### para el inversor)

Volatilidad = desviación de los
retornos diarios por mes.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/nvidia_bolsa.csv'
)

df['mes'] = df['fecha'].str[:7]

r = (
  df
  .groupby('mes')['retorno_pct']
  .std()
  .round(2)
  .sort_values(ascending=False)
)

print('Top meses más volátiles:')
print(r.head(5))
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Data Center ya pesa
### más que Gaming en
### los ingresos?

El giro a la IA. Comparamos el
último trimestre por segmento.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/nvidia_bolsa.csv'
)

cols = [
  c for c in df.columns
  if c.startswith('ing_')
]

ult = df['trimestre'].iloc[-1]
fila = (
  df[df['trimestre'] == ult]
  [cols].iloc[0]
)

pct = (fila / fila.sum() * 100).round(1)

tabla = pd.DataFrame({
  'ingresos_M': fila,
  '%': pct,
}).sort_values('%', ascending=False)

print(f'Ingresos {ult} por segmento:')
print(tabla)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuánto creció Data
### Center trimestre a
### trimestre? (QoQ)

El motor del boom. Miramos su
crecimiento intertrimestral.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/nvidia_bolsa.csv'
)

s = (
  df
  .groupby('trimestre')
  ['ing_data_center_M']
  .first()
  .sort_index()
)

tabla = pd.DataFrame({
  'ingresos_M': s.round(0),
  'qoq_%': (
    s.pct_change() * 100
  ).round(1),
})

print('Data Center QoQ:')
print(tabla)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿El precio sube cuando
### sube el volumen?
### (correlación)

¿El dinero que entra empuja la
cotización? Medimos correlación.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/nvidia_bolsa.csv'
)

df['mes'] = df['fecha'].str[:7]

m = (
  df
  .groupby('mes')
  .agg(
    precio=('precio_accion', 'mean'),
    vol=('volumen', 'mean'),
  )
)

corr = m['precio'].corr(m['vol'])

print('Corr precio vs volumen:')
print(round(corr, 3))
```
