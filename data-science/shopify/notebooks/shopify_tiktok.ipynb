### 💼 PREGUNTA ENTREVISTA:
### ¿Qué categoría genera
### más ingreso bruto?
### (pedidos x ticket)

Ingreso = pedidos x ticket medio.
Es lo que mueve la caja.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/shopify_pedidos.csv'
)

df['ingreso'] = (
  df['pedidos']
  * df['ticket_medio_usd']
)

r = (
  df
  .groupby('categoria')['ingreso']
  .sum()
  .sort_values(ascending=False)
)

pct = (r / r.sum() * 100).round(1)

tabla = pd.DataFrame({
  'ingreso_usd': r.round(0),
  '%': pct,
})

print('Ingreso por categoría:')
print(tabla)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Desde qué canal
### gastan más por
### pedido los clientes?

Ticket medio por canal. ¿Web, móvil,
redes o tienda física (POS)?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/shopify_pedidos.csv'
)

r = (
  df
  .groupby('canal')
  .agg(
    ticket=('ticket_medio_usd',
            'mean'),
    pedidos=('pedidos', 'sum'),
  )
  .round(2)
  .sort_values('ticket',
    ascending=False)
)

print('Ticket y pedidos por canal:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué categoría nos
### come margen con
### devoluciones?

Tasa devolución = devoluciones /
pedidos. Devolver cuesta dinero.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/shopify_pedidos.csv'
)

g = (
  df
  .groupby('categoria')
  [['devoluciones', 'pedidos']]
  .sum()
)

g['tasa_dev_%'] = (
  g['devoluciones']
  / g['pedidos'] * 100
).round(2)

print('Tasa devolución por categoría:')
print(
  g['tasa_dev_%']
  .sort_values(ascending=False)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué países son el
### mejor mercado por
### ingreso neto?

Neto = (pedidos - devoluciones) x
ticket. Descontamos lo devuelto.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/shopify_pedidos.csv'
)

df['neto'] = (
  (df['pedidos']
   - df['devoluciones'])
  * df['ticket_medio_usd']
)

r = (
  df
  .groupby('pais')['neto']
  .sum()
  .sort_values(ascending=False)
  .round(0)
)

print('Ingreso neto por país:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuándo es el pico
### de ventas del año?
### (campaña Q4)

Ingreso por mes para detectar el
Black Friday / Navidad..

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/shopify_pedidos.csv'
)

df['ingreso'] = (
  df['pedidos']
  * df['ticket_medio_usd']
)

s = (
  df
  .groupby('mes')['ingreso']
  .sum()
)

q4 = s.loc[[10, 11, 12]].mean()
resto = s.drop([10, 11, 12]).mean()
lift = (q4 / resto - 1) * 100

print('Ingreso por mes:')
print(s.round(0))
print(f'\nQ4 vs resto: +{lift:.1f}%')
```
