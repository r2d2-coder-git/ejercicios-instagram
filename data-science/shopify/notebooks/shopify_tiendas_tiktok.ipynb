### 💼 PREGUNTA ENTREVISTA:
### ¿Qué plan concentra
### el GMV y cuánto
### aporta cada tienda?

GMV total y medio por plan. Donde
está el dinero del marketplace.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/'
  'shopify_tiendas.csv'
)

g = (
  df
  .groupby('plan')
  .agg(
    gmv_total=('gmv_usd',
               'sum'),
    gmv_medio=('gmv_usd',
               'mean'),
    n=('gmv_usd', 'size'),
  )
  .round(0)
)

g['%'] = (
  g['gmv_total']
  / g['gmv_total'].sum() * 100
).round(1)

print('GMV por plan:')
print(
  g.sort_values('gmv_total',
    ascending=False)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué plan tiene peor
### retención de tiendas
### activas? (churn)

% de tiendas inactivas por plan.
La fuga silenciosa de ingresos.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/'
  'shopify_tiendas.csv'
)

r = (
  df
  .groupby('plan')['activa']
  .agg(['mean', 'size'])
)

r['churn_%'] = (
  (1 - r['mean']) * 100
).round(1)

r['activa_%'] = (
  r['mean'] * 100
).round(1)

print('Retención por plan:')
print(
  r[['activa_%', 'churn_%',
     'size']]
  .sort_values('churn_%',
    ascending=False)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Instalar más apps se
### asocia a más GMV?
### (correlación)

Correlación apps vs GMV. ¿El
ecosistema de apps mueve ventas?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/'
  'shopify_tiendas.csv'
)

corr = (
  df['apps_instaladas']
  .corr(df['gmv_usd'])
)

r = (
  df
  .groupby('apps_instaladas')
  ['gmv_usd']
  .mean()
  .round(0)
  .head(8)
)

print(f'Correlación: {corr:.3f}')
print('\nGMV medio por nº apps:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué país tiene las
### tiendas con mayor
### GMV por pedido?

GMV / pedidos = valor por pedido.
El poder de compra por mercado.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/'
  'shopify_tiendas.csv'
)

g = (
  df
  .groupby('pais')
  [['gmv_usd', 'pedidos_mes']]
  .sum()
)

g['gmv_pedido'] = (
  g['gmv_usd']
  / g['pedidos_mes']
).round(2)

print('GMV por pedido / país:')
print(
  g['gmv_pedido']
  .sort_values(ascending=False)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Las tiendas nuevas
### aguantan activas más
### que las veteranas?

Tasa de actividad por tramo de
antigüedad. El ciclo de vida real.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/'
  'shopify_tiendas.csv'
)

df['antig'] = pd.cut(
  df['antiguedad_meses'],
  bins=[0, 12, 36, 60, 999],
  labels=['<1a', '1-3a',
          '3-5a', '+5a'],
)

r = (
  df
  .groupby('antig',
    observed=True)['activa']
  .mean()
  .mul(100)
  .round(1)
)

print('% activas por antigüedad:')
print(r)
```
