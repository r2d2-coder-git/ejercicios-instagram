### 💼 PREGUNTA ENTREVISTA:
### ¿Qué parque ingresa
### más y con qué precio
### medio de entrada?

Ingresos = visitantes x precio.
¿Volumen o precio premium?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/disney_parques.csv'
)

r = (
  df
  .groupby('parque')
  .agg(
    ingresos=('ingresos_usd', 'sum'),
    precio=('precio_entrada_usd',
      'mean'),
    visit=('visitantes', 'mean'),
  )
  .round(0)
  .sort_values('ingresos',
    ascending=False)
)

print('Ingresos por parque:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿La masificación mata
### la satisfacción?
### (ocupación vs nota)

Troceamos la ocupación y vemos
la nota media de cada tramo.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/disney_parques.csv'
)

df['tramo'] = pd.cut(
  df['ocupacion_pct'],
  [0, 40, 60, 80, 100],
  labels=['Baja', 'Media',
    'Alta', 'Llena'],
)

r = (
  df
  .groupby('tramo', observed=True)
  ['satisfaccion']
  .mean()
  .round(2)
)

print('Satisfacción por ocupación:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuánto más caro es
### el finde frente a
### un día de diario?

Comparamos precio e ingresos entre
finde y laborable.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/disney_parques.csv'
)

df['tipo'] = df['finde'].map(
  {0: 'Diario', 1: 'Finde'}
)

g = (
  df
  .groupby('tipo')
  .agg(
    precio=('precio_entrada_usd',
      'mean'),
    visit=('visitantes', 'mean'),
  )
  .round(1)
)

dif = (
  g.loc['Finde', 'precio']
  / g.loc['Diario', 'precio'] - 1
) * 100

print(g)
print(f'\nFinde +{dif:.1f}% precio')
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué meses concentran
### las visitas?
### (estacionalidad)

Visitantes medios por mes para
planificar personal y precios.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/disney_parques.csv'
)

g = (
  df
  .groupby('mes')['visitantes']
  .mean()
)

idx = (g / g.mean() * 100).round(0)

tabla = pd.DataFrame({
  'visit_medios': g.round(0),
  'indice': idx,
}).sort_values('indice',
  ascending=False)

print('Estacionalidad por mes:')
print(tabla.head(6))
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué país tiene los
### parques mejor
### valorados?

Nota media ponderada por país
para comparar la experiencia.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/disney_parques.csv'
)

r = (
  df
  .groupby('pais')
  .agg(
    satis=('satisfaccion', 'mean'),
    ocup=('ocupacion_pct', 'mean'),
    n=('fecha', 'count'),
  )
  .round(2)
  .sort_values('satis',
    ascending=False)
)

print('Satisfacción por país:')
print(r)
```
