### 💼 PREGUNTA ENTREVISTA:
### ¿Qué franquicia manda
### en horas vistas
### y qué % del total?

Las horas vistas son el KPI estrella
del streaming (más que suscriptores).

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/disney_plus.csv'
)

total = df['horas_vistas'].sum()

r = (
  df
  .groupby('franquicia')
  ['horas_vistas']
  .sum()
  .sort_values(ascending=False)
)

tabla = pd.DataFrame({
  'horas_M': (r / 1e6).round(1),
  '%': (r / total * 100).round(1),
})

print('Horas por franquicia:')
print(tabla)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué títulos enganchan
### más? (horas por
### espectador)

Horas / espectador = cuánto se queda
la gente. Mide retención real.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/disney_plus.csv'
)

g = (
  df
  .groupby('titulo')
  .agg(
    horas=('horas_vistas', 'sum'),
    espec=('espectadores', 'sum'),
  )
)

g['horas_x_espec'] = (
  g['horas'] / g['espec']
).round(2)

print('Top enganche:')
print(
  g.sort_values('horas_x_espec',
    ascending=False)
  ['horas_x_espec']
  .head(6)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué región consume
### más y en cuál
### crecemos poco?

Comparamos horas por región para
saber dónde invertir en contenido.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/disney_plus.csv'
)

r = (
  df
  .groupby('region')
  .agg(
    horas=('horas_vistas', 'sum'),
    compl=('completado_pct',
           'mean'),
  )
  .sort_values('horas',
    ascending=False)
)

r['horas_M'] = (
  r['horas'] / 1e6
).round(1)
r['compl'] = r['compl'].round(1)

print('Consumo por región:')
print(r[['horas_M', 'compl']])
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Pagar presupuestazos
### garantiza más
### audiencia?

Correlación presupuesto vs horas.
¿El dinero compra espectadores?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/disney_plus.csv'
)

g = (
  df
  .groupby('titulo')
  .agg(
    presup=('presupuesto_m_usd',
            'first'),
    horas=('horas_vistas', 'sum'),
    imdb=('imdb', 'first'),
  )
)

c1 = g['presup'].corr(g['horas'])
c2 = g['imdb'].corr(g['horas'])

print(f'Corr presup-horas: {c1:.2f}')
print(f'Corr imdb-horas:   {c2:.2f}')
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Series o pelis dan
### más horas por
### dólar invertido?

Eficiencia = horas vistas por cada
millón de presupuesto gastado.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/disney_plus.csv'
)

g = (
  df
  .groupby(['tipo', 'titulo'])
  .agg(
    horas=('horas_vistas', 'sum'),
    presup=('presupuesto_m_usd',
            'first'),
  )
  .reset_index()
)

g['horas_x_M'] = (
  g['horas'] / g['presup']
)

r = (
  g
  .groupby('tipo')['horas_x_M']
  .mean()
  .round(0)
  .sort_values(ascending=False)
)

print('Horas por M$ (tipo):')
print(r)
```
