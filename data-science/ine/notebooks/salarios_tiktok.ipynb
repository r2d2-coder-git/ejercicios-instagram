### 💼 PREGUNTA ENTREVISTA:
### ¿Cuál es la brecha salarial
### entre hombres y mujeres?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/salarios_espana.csv'
)

result = (
  df
  .groupby('sexo')
  ['salario_bruto_anual']
  .median()
)

brecha = (
  1 - result['Mujer']
  / result['Hombre']
) * 100

print("Salario mediano anual:")
print(result.round(0))
print(f"\nBrecha: {brecha:.1f}%")
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuánto más cobra
### un indefinido que un temporal?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/salarios_espana.csv'
)

result = (
  df
  .groupby('tipo_contrato')
  ['salario_bruto_anual']
  .agg(['median', 'mean', 'count'])
)

diff = (
  result.loc['Indefinido','median']
  - result.loc['Temporal','median']
)

print(result.round(0))
print(f"\nDiferencia: {diff:,.0f}€")
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuánto sube el salario
### por cada nivel de estudios?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/salarios_espana.csv'
)

order = [
  'Sin estudios', 'Primaria',
  'ESO', 'FP medio',
  'Bachiller', 'FP superior',
  'Universidad'
]

result = (
  df
  .groupby('estudios')
  ['salario_bruto_anual']
  .median()
  .reindex(order)
)

print("Salario mediano por estudios:")
print(result.round(0))

ratio = (
  result['Universidad']
  / result['ESO']
)
print(f"\nUni cobra {ratio:.1f}x más")
print("que alguien con ESO")
```

### 💼 PREGUNTA ENTREVISTA:
### ¿El salario sube con
### la antigüedad? ¿Cuánto?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/salarios_espana.csv'
)

bins = [0, 1, 3, 5, 10, 20, 50]
labels = [
  '<1', '1-3', '3-5',
  '5-10', '10-20', '20+'
]

df['rango_anti'] = pd.cut(
  df['antiguedad_anos'],
  bins=bins, labels=labels
)

result = (
  df
  .groupby('rango_anti')
  ['salario_bruto_anual']
  .median()
)

print("Salario por antigüedad:")
print(result.round(0))

subida = (
  result['20+'] / result['<1']
  - 1
) * 100
print(f"\n+{subida:.0f}% con 20+ años")
```

### 💼 PREGUNTA ENTREVISTA:
### ¿La brecha salarial
### existe en todos los
### niveles de estudios?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/salarios_espana.csv'
)

result = (
  df
  .groupby(['estudios', 'sexo'])
  ['salario_bruto_anual']
  .median()
  .unstack()
)

result['brecha_%'] = (
  (1 - result['Mujer']
   / result['Hombre']) * 100
)

print(result[['brecha_%']]
  .sort_values('brecha_%',
    ascending=False)
  .round(1)
)
```
