### 💼 PREGUNTA ENTREVISTA:
### ¿Qué % de España
### es población extranjera
### y cómo ha crecido?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/evolucion_migracion.csv',
  sep=';'
)

df['Total'] = pd.to_numeric(
  df['Total']
  .str.replace('.','', regex=False),
  errors='coerce'
)

mask = (
  (df['Grupo quinquenal de edad']
   == 'Todas las edades')
  & (df['Sexo'] == 'Total')
  & (df['Periodo'].str.contains(
    'enero'))
)

t = df[mask].copy()

total = (
  t[t['Nacionalidad']=='Total']
  .set_index('Periodo')['Total']
)
extr = (
  t[t['Nacionalidad']=='Extranjera']
  .set_index('Periodo')['Total']
)

pct = (extr / total * 100).dropna()

print("% población extranjera:")
print(f"2002: {pct.iloc[-1]:.1f}%")
print(f"2010: {pct.iloc[-9]:.1f}%")
print(f"2025: {pct.iloc[0]:.1f}%")
print()
print(f"Extranjeros 2025:")
print(f"{int(extr.iloc[0]):,}")
print("de", f"{int(total.iloc[0]):,}")
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué nacionalidades
### han EXPLOTADO más
### en 10 años?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/evolucion_migracion.csv',
  sep=';'
)

df['Total'] = pd.to_numeric(
  df['Total']
  .str.replace('.','', regex=False),
  errors='coerce'
)

mask = (
  (df['Grupo quinquenal de edad']
   == 'Todas las edades')
  & (df['Sexo'] == 'Total')
)
t = df[mask].copy()

# Excluir agrupaciones
excl = [
  'Total', 'Española', 'Extranjera',
  'País de la UE27_2020 sin España',
  'País de la UE28 sin España',
  'País de Europa menos UE27_2020',
  'País de Europa menos UE28',
  'De Africa', 'De América del Norte',
  'De Centro América y Caribe',
  'De Sudamérica', 'De Asia',
  'De Oceanía', 'Apátridas',
  'Otro país de la Unión Europea sin España',
  'Otro país del resto de Europa',
  'Otro país de África',
  'Otros países de Centro América y Caribe',
  'Otro país de Sudamérica',
  'Otro país de Asia'
]

paises = t[~t['Nacionalidad'].isin(excl)]

ene15 = (
  paises[
    paises['Periodo']
    == '1 de enero de 2015'
  ].set_index('Nacionalidad')['Total']
)

ene25 = (
  paises[
    paises['Periodo']
    == '1 de enero de 2025'
  ].set_index('Nacionalidad')['Total']
)

crec = (
  ((ene25 / ene15 - 1) * 100)
  .dropna()
  .sort_values(ascending=False)
  .head(10)
)

print("Crecimiento 2015-2025:")
print(crec.round(0).astype(int)
  .to_string())
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Cuáles son las 15
### nacionalidades con más
### población en España?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/evolucion_migracion.csv',
  sep=';'
)

df['Total'] = pd.to_numeric(
  df['Total']
  .str.replace('.','', regex=False),
  errors='coerce'
)

mask = (
  (df['Grupo quinquenal de edad']
   == 'Todas las edades')
  & (df['Sexo'] == 'Total')
  & (df['Periodo']
     == '1 de enero de 2025')
)

t = df[mask].copy()

excl = [
  'Total', 'Española', 'Extranjera',
  'País de la UE27_2020 sin España',
  'País de la UE28 sin España',
  'País de Europa menos UE27_2020',
  'País de Europa menos UE28',
  'De Africa', 'De América del Norte',
  'De Centro América y Caribe',
  'De Sudamérica', 'De Asia',
  'De Oceanía', 'Apátridas',
  'Otro país de la Unión Europea sin España',
  'Otro país del resto de Europa',
  'Otro país de África',
  'Otros países de Centro América y Caribe',
  'Otro país de Sudamérica',
  'Otro país de Asia'
]

result = (
  t[~t['Nacionalidad'].isin(excl)]
  .sort_values('Total',
    ascending=False)
  .head(15)
  [['Nacionalidad', 'Total']]
)

total_ext = t[
  t['Nacionalidad']=='Extranjera'
]['Total'].values[0]

result['%'] = (
  result['Total'] / total_ext * 100
)

print("Top 15 nacionalidades 2025:")
print(
  result.to_string(index=False)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿De Marruecos vienen más
### hombres o mujeres?
### ¿Y de Colombia?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/evolucion_migracion.csv',
  sep=';'
)

df['Total'] = pd.to_numeric(
  df['Total']
  .str.replace('.','', regex=False),
  errors='coerce'
)

mask = (
  (df['Grupo quinquenal de edad']
   == 'Todas las edades')
  & (df['Sexo'] != 'Total')
  & (df['Periodo']
     == '1 de enero de 2025')
)

t = df[mask].copy()

paises = [
  'Marruecos', 'Colombia',
  'Rumanía', 'Venezuela',
  'China', 'Honduras'
]

result = (
  t[t['Nacionalidad'].isin(paises)]
  .pivot_table(
    index='Nacionalidad',
    columns='Sexo',
    values='Total'
  )
)

result['%_hombres'] = (
  result['Hombres']
  / (result['Hombres']
     + result['Mujeres'])
  * 100
)

print("Ratio hombres/mujeres 2025:")
print(
  result[['%_hombres']]
  .sort_values('%_hombres',
    ascending=False)
  .round(1)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿En qué año llegó más
### gente de Venezuela
### a España?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/evolucion_migracion.csv',
  sep=';'
)

df['Total'] = pd.to_numeric(
  df['Total']
  .str.replace('.','', regex=False),
  errors='coerce'
)

mask = (
  (df['Grupo quinquenal de edad']
   == 'Todas las edades')
  & (df['Sexo'] == 'Total')
  & (df['Nacionalidad']=='Venezuela')
  & (df['Periodo'].str.contains(
    'enero'))
)

ven = (
  df[mask]
  .sort_values('Periodo')
  .copy()
)

ven['incremento'] = ven['Total'].diff()

ven['year'] = (
  ven['Periodo']
  .str.extract(r'(\d{4})')
)

print("Venezolanos en España:")
print(
  ven[['year','Total','incremento']]
  .tail(10)
  .to_string(index=False)
)

pico = ven.loc[
  ven['incremento'].idxmax()
]
print(f"\nMáx llegada: {pico['year']}")
print(f"+{int(pico['incremento']):,}")
```
