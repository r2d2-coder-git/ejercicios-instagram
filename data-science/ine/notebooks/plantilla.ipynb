### 💼 PREGUNTA ENTREVISTA:
### ¿Cuánto es el salario
### por cada nivel de estudios?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/salarios_espana.csv'
)

order = [
  'Sin estudios', 'Primaria',
  'ESO', 'FP medio',
  'Bachiller', 'FP superior',
  'Universidad'
]

result = (
    df.
    groupby('estudios')
    ['salario_bruto_anual']
    .mean()
    .reindex(order)
)

print("Salario medio por estudios: ")
print(result.round(0))
```
