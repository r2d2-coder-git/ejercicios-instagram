### ¿Más MAYÚSCULAS = más views?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/youtube_trending_us.csv'
)

bins = [0, 20, 40, 60, 80, 100]
labels = [
  '0-20%', '20-40%',
  '40-60%', '60-80%', '80-100%'
]

# % de letras en mayúsculas 
df['caps_pct'] = (
    df['title'].str.count('[A-Z]')
    / df['title'].str.len() * 100
)

# Separar los datos por intervalos
df['caps_bin'] = pd.cut(
    df['caps_pct'],
    bins = bins, labels = labels
)

# Calcular media de visitas por intervalo
result = (
    df.
    groupby('caps_bin',
    observed = False)
    ['views']
    .agg(['median', 'count'])
)

result['median_M'] = (
    result['median'] /1e6
)

print("views por % mayus:")
print(result[['median_M', 'count']].round(2))
```
