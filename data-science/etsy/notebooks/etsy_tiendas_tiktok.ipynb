### 💼 PREGUNTA ENTREVISTA:
### ¿Qué categoría factura
### más y con qué ticket
### medio por venta?

Ingresos totales y precio medio.
No es lo mismo volumen que valor.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/etsy_tiendas.csv'
)

g = (
  df
  .groupby('categoria_principal')
  [['ingresos_usd',
    'ventas_totales']]
  .sum()
)

g['ticket'] = (
  g['ingresos_usd']
  / g['ventas_totales']
).round(2)

print('Ingresos y ticket:')
print(
  g.sort_values(
    'ingresos_usd',
    ascending=False)
  .round(0)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Las tiendas veteranas
### venden más que las
### nuevas de verdad?

Agrupamos por tramos de antigüedad
y comparamos ventas medias.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/etsy_tiendas.csv'
)

df['antig'] = pd.cut(
  df['antiguedad_meses'],
  bins=[0, 12, 36, 60, 999],
  labels=['<1a', '1-3a',
          '3-5a', '+5a'],
)

r = (
  df
  .groupby('antig',
    observed=True)
  .agg(
    ventas=('ventas_totales',
            'mean'),
    ingresos=('ingresos_usd',
              'mean'),
    n=('ventas_totales',
       'size'),
  )
  .round(0)
)

print('Por antigüedad de tienda:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Un catálogo grande
### garantiza más
### ingresos? (correlación)

Correlación productos vs ingresos.
¿Más SKUs = más caja?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/etsy_tiendas.csv'
)

corr = (
  df['productos']
  .corr(df['ingresos_usd'])
)

df['tramo'] = pd.qcut(
  df['productos'], 4,
  labels=['pocos', 'medio',
          'muchos', 'enorme'],
)

r = (
  df
  .groupby('tramo',
    observed=True)
  ['ingresos_usd']
  .mean()
  .round(0)
)

print(f'Correlación: {corr:.3f}')
print('\nIngresos por catálogo:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué país tiene las
### tiendas más rentables
### por reseña?

Ingreso por reseña: proxy de valor
generado por cada cliente feliz.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/etsy_tiendas.csv'
)

g = (
  df
  .groupby('pais')
  [['ingresos_usd', 'resenas']]
  .sum()
)

g['usd_resena'] = (
  g['ingresos_usd']
  / g['resenas']
).round(2)

print('Ingreso por reseña:')
print(
  g['usd_resena']
  .sort_values(ascending=False)
)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué tiendas son las
### top 10% y cuánto
### pesan en ingresos?

Concentración de ingresos en la
élite de vendedores (Pareto).

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/etsy_tiendas.csv'
)

s = (
  df['ingresos_usd']
  .sort_values(ascending=False)
  .reset_index(drop=True)
)

top = int(len(s) * 0.1)
pct = (
  s.head(top).sum()
  / s.sum() * 100
)

print(f'Tiendas: {len(s)}')
print(f'Top 10%: {top}')
print(
  f'Generan: {pct:.1f}% '
  'de ingresos'
)
```
