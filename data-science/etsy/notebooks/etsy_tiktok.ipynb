### 💼 PREGUNTA ENTREVISTA:
### ¿Qué categoría factura
### más? (precio x
### ventas = GMV)

GMV = volumen de negocio. Precio alto
no sirve si no hay ventas.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/etsy_productos.csv'
)

df['gmv'] = (
  df['precio_usd'] * df['ventas']
)

r = (
  df
  .groupby('categoria')['gmv']
  .sum()
  .sort_values(ascending=False)
)

pct = (r / r.sum() * 100).round(1)

tabla = pd.DataFrame({
  'gmv_usd': r.round(0),
  '%': pct,
})

print('GMV por categoría:')
print(tabla)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Lo hecho a mano se
### vende más caro
### y gusta más?

El sello artesano es la marca de Etsy.
¿Se traduce en precio y rating?

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/etsy_productos.csv'
)

r = (
  df
  .groupby('hecho_a_mano')
  .agg(
    precio=('precio_usd',
            'median'),
    rating=('rating', 'mean'),
    ventas=('ventas', 'median'),
  )
  .round(2)
)

r.index = r.index.map(
  {True: 'Hecho a mano',
   False: 'Fabricado'})

print('Artesano vs fabricado:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Qué país vendedor
### domina el marketplace
### y con qué ticket?

Volumen de ventas y precio medio por
país de origen del vendedor.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/etsy_productos.csv'
)

r = (
  df
  .groupby('pais_vendedor')
  .agg(
    productos=('ventas', 'size'),
    ventas=('ventas', 'sum'),
    precio=('precio_usd',
            'median'),
  )
  .round(2)
  .sort_values('ventas',
    ascending=False)
)

print('Por país vendedor:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Precio y rating están
### reñidos? ¿Pagar más
### da mejor producto?

Partimos el precio en cuartiles y
miramos el rating medio de cada uno.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/etsy_productos.csv'
)

df['q_precio'] = pd.qcut(
  df['precio_usd'], 4,
  labels=['Barato', 'Medio',
          'Caro', 'Premium'],
)

r = (
  df
  .groupby('q_precio',
    observed=True)
  .agg(
    rating=('rating', 'mean'),
    ventas=('ventas', 'median'),
  )
  .round(2)
)

print('Rating por nivel de precio:')
print(r)
```

### 💼 PREGUNTA ENTREVISTA:
### ¿Enviar rápido se
### traduce en más
### ventas?

Agrupamos por rapidez de envío y
vemos ventas medianas por tramo.

```python
import pandas as pd

df = pd.read_csv(
  '../datasets/etsy_productos.csv'
)

df['velocidad'] = pd.cut(
  df['envio_dias'],
  bins=[0, 3, 7, 100],
  labels=['Rápido (<=3d)',
          'Normal (4-7d)',
          'Lento (>7d)'],
)

r = (
  df
  .groupby('velocidad',
    observed=True)['ventas']
  .median()
  .round(0)
)

print('Ventas medianas por envío:')
print(r)
```
